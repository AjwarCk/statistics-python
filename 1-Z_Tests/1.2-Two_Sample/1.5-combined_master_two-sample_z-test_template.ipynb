#### Master Function for Two-Sample Z-Tests

1. Function Signature:
Create a function like `two_sample_z_test()` that takes in the two group parameters, the significance level, and the type of tail (right, left, two).

2. Inside the function:
Calculate the difference in means, the combined standard error, the Z-statics, and then choose the correct p-value calculation and critical value based on the tail type.

3. Output:
It will print out difference, SE, Z-stat, p-value, critical value and the final decision. Basically, it will look just like the one sample master but adapted the two sample scenario.

```python
import numpy as np
from scipy import stats

def two_sample_z_test(mean_A, sigma_A, n_A, mean_B, sigma_B, n_B, alpha, tail_type):
    """
    Perform a two-sample Z-Test.
    tail_type: 'right', 'left', or 'two' for the direction of the test.
    """
    
    # Calculate the difference in means
    diff = mean_B - mean_A

    # Calculate standard error for two-sample Z-test
    SE = np.sqrt((sigma_A**2/n_A)+(sigma_B**2/n_B))

    # Calculate Z-statistic
    z_stat = diff/SE

    # Determine p-value based on the tail type
    if tail_type == 'right':
        p_value = 1 - stats.norm.cdf(z_stat)
        crit_value = stats.norm.ppf(1 - alpha)
    elif tail_type == 'left':
        p_value = stats.norm.cdf(z_stat)
        crit_value = stats.norm.ppf(alpha)
    elif tail_type == 'two':
        p_value = 2 * (1 - stats.norm.cdf(z_stat))
        crit_value = stats.norm.ppf(1-alpha/2) # For two-tailed we have +/- crit

    # Output result
    print(f"Difference (B - A): {diff:.4f}")
    print(f"Standard Error (SE): {SE:.4f}")
    print(f"Z_Statistic: {z_stat:.4f}")
    print(f"P-Value: {p_value:.4f}")

    if tail_type == 'two':
        print(f"Critical Values: Left = {-crit_value:.4f}, Right={crit_value:.4f}")
    else:
        print(f"Critical Value: {crit_value:.4f}")

    # Decision
    if (tail_type == 'right' and z_stat > crit_value) or \
        (tail_type == 'left' and z_stat < crit_value) or \
        (tail_type == 'two' and (abs(z_stat) > crit_value)):
        print("Reject H0: Statistically significant difference detected.")
    else:
        print("Fail to Reject H0: No statistically significant difference.")

# Example
two_sample_z_test(
    mean_A=1250, sigma_A=300, n_A=500,
    mean_B=1320, sigma_B=320, n_B=520,
    alpha=0.05, tail_type='two'
)
```
