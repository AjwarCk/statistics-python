#### Left-Tailed - Two-Sample Z-Test for Proportions

**1 - What is it?**

A **two-tailed two-sample Z-test for proportions** checks whether the **proportion** (rate/percentage) of Group B is significantly LOWER than the proportion of Group A.

Use this when:

- Outcome is binary (convert/not convert, defect/no defect)
- Comparing **two independent groups**
- Sample sizes are large
- You specifically want to detect a **decrease**

`Think of it like:`

"Did the new version reduce conversion compared to the old version?"

**2 - Real-Life Example**

**Scenario:**

After releasing a new checkout flow, stakeholders worry:

    "The new checkout might have reduced conversion rate."

Run an A/B test:

**Group A (Control - old checkout)**

- Visitors = 8,000
- Conversions = 320
- Conversion rate = 4.00%

**Group B (Treatment - new checkout)**

- Visitors = 7,500
- Conversion = 255
- Conversion rate = 3.40%

Question:

*Is Group B's conversion rate significantly LOWER than Group A's*

This is a left-tailed two-sample Z-test for proportions.

**4 - Formula**

Sample proportions:

- P^A = xA/nA
- P^B = xB/nB

Pooled proportion (MANDATORY)

Under H0, proportions are equal:

    P^pooled = xA+xB / nA+nB
