#### Left-Tailed - One-Sample Z-Test for Proportions

**1 - What is it?**

A `left-tailed one-sample Z-test for proportions` checks whether a sample proportion is significantly **LOWER** than known population (benchmark) proportion.

Use this when:
- Outcome is binary (yes/no, success/failure)
- Comparing one sample vs fixed benchmark
- Sample size is large
- Testing for a decrease only

`Think of it like:`

"Did my observed conversion rate drop below the expected baseline, or is this just random variation?"

**2 - Real-Life Example**

**Scenario:**

Operation team is worried:
    "The recent website changes might have reduced the checkout conversion rate below our usual 4% benchmark."

Collect data:

- Total visitors = 1,200
- Conversions = 36
- Observed conversion = 3%
- Benchmark (population proportion) = 4%

Question:

`Is 3% significantly lower than 4%?`

This is a `left-tailed one-sample Z-test for proportions.`

**3 - Hypotheses**

Let:

- p0 = benchmark proportion
- p^ = sample proportion

**Hypotheses (left-tailed)**
- H0 (Null): p = p0 = 0.04
- H1 (Alternate): p < 0.04

Left-tailed because we care **only about a decrease.**

**4 - Formula**

Sample proportion = p^ = conversion / sample size = x/n

Standard Error (IMPORTANT):

    SE = SQRT(p0*(1-p0)/n)

`Critical rule:` For **One-sample proportion tests,** SE is calculated using p0, not p^.

Z-Statistics = p^ - p0 / SE

Left-tailed p-value = p = stats.norm.cdf(z)

**Parameter Used**

- Conversions x = 36
- Sample size n = 1200
- Benchmark p0 = 0.04
- Significance level a = 0.05

**5 - Manual Calculation (Step-by-Step)**

Step1: Sample proportion :

    p^ = 36/1200 = 0.03

Step2: Standard Error : 

    SE = SQRT(0.04*(1-0.04)/1200) = SQRT(0.000032) = 0.00566

Step3: Z-Statistic : 

    Z = 0.03 - 0.04 / 0.00566 = -1.77

Step4: p-value (left-tailed)

    p = stats.norm.cdf(-1.77) = 0.038

**6 - Python Implementation (Step-by-Step)**

```python
import numpy as np
from scipy import stats

# -------------------------------------------------------------
# Step 1: Inputs
# -------------------------------------------------------------
x = 36              # number of conversions
n = 1200            # total visitors
p0 = 0.04           # benchmark proportion
alpha = 0.05        # left-tailed test significance value

# -------------------------------------------------------------
# Step 2: Sample proportion
# -------------------------------------------------------------
p_hat = x/n
print(f"Sample Proportion (p^): {p_hat:.4f}")

# -------------------------------------------------------------
# Step 3: Standard Error (use p0)
# -------------------------------------------------------------
se = np.sqrt((p0*(1-p0))/n)
print(f"Standard Error (SE): {se:.4f}")

# -------------------------------------------------------------
# Step 4: Z-Statistic
# -------------------------------------------------------------
z_stat = (p_hat - p0) / se
print(f"Z Statistics: {z_stat:.4f}")

# -------------------------------------------------------------
# Step 5: P-Value (left-tailed)
# -------------------------------------------------------------
p_value = stats.norm.cdf(z_stat)
print(f"P-Value: {p_value:.6f}")

# -------------------------------------------------------------
# Step 6: Critical Value
# -------------------------------------------------------------
critical_value = stats.norm.ppf(alpha)
print(f"Crictical Value: {critical_value:.4f}")

# -------------------------------------------------------------
# Step 7: Decision
# -------------------------------------------------------------
print("\n------------ Decision --------------")
if p_value < alpha:
    print("Reject H0: Conversion rate is significantly lower than benchmark.")
else:
    print("Fail to Reject H0: No significant decrease detected.")
```

**7 - Business Interpretation**

- Observed conversion rate = 3%
- Benchmark = 4%
- Z = 1.77 -> 1.77 standard errors below expection
- p = 0.0038 -> 3.8% chance this drop is random

**Final Conclusion**

The conversion rate is statistically significant lower than the benchmark.
The drop is real and should be investigated.
